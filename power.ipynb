```python

total_revenue = df['total_price'].sum()
total_pizza_sold = df['quantity'].sum()
total_orders = df['order_id'].nunique()

avg_order_value = total_revenue / total_orders
avg_pizza_per_order = total_pizza_sold / total_orders

print("The total revenue of the dataset:", total_revenue)
print("The total number of pizzas sold:", total_pizza_sold)
print("The total number of orders placed:", total_orders)
print("The average order value:", avg_order_value)
print("The average number of pizzas per order:", avg_pizza_per_order)
```

```python
ingredients = (
    df['pizza_ingredients']
    .str.split(',')
    .explode()
    .str.strip()
    .value_counts()
    .reset_index()
    .rename(columns={'index': 'ingredient', 'pizza_ingredients': 'counts'})
     #his will split the ingredients column by comma and then explode the list into separate rows and then count the values of each ingredient
)

print(ingredients.head(15)) # to get the top 15 ingredients used in the pizzas
```

```python

# Convert order_date into datetime format
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

# Extract day name from order_date
df['day_name'] = df['order_date'].dt.day_name()

# Arrange days in weekly order
weekly_orders = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

df['day_name'] = pd.Categorical(
    df['day_name'],
    categories=weekly_orders,
    ordered=True
)

# Count unique orders for each day
orders_by_day = df.groupby('day_name', observed=False)['order_id'].nunique()

# Create bar chart
ax = orders_by_day.plot(
    kind='bar',
    figsize=(10, 5),
    color='orange',
    edgecolor='black'
)

# Add chart title and labels
plt.title('Number of Orders by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Number of Orders')
plt.xticks(rotation=45)

# Display values on top of each bar
for i, value in enumerate(orders_by_day):
    plt.text(
        i, value + 20, str(value),
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

```python


# Convert order_date into datetime format
df['order_date'] = pd.to_datetime(df['order_date'], dayfirst=True)

# Extract day name from order_date
df['day_name'] = df['order_date'].dt.day_name()

# Arrange days in weekly order
weekly_orders = [
    "Monday", "Tuesday", "Wednesday", "Thursday",
    "Friday", "Saturday", "Sunday"
]

df['day_name'] = pd.Categorical(
    df['day_name'],
    categories=weekly_orders,
    ordered=True
)

# Count unique orders for each day
orders_by_day = df.groupby('day_name', observed=False)['total_price'].sum()  # Changed from 'order_id' to 'total_price' to sum the total price for each day

# Create bar chart
ax = orders_by_day.plot(
    kind='bar',
    figsize=(10, 5),
    color='purple',
    edgecolor='black'
)

# Add chart title and labels
plt.title('Number of Orders by Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('totl rev' \
'f Orders')
plt.xticks(rotation=45)

# Display values on top of each bar
for i, value in enumerate(orders_by_day):
    plt.text(
        i, value + 20, str(value),
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

```python

df['order_time'] = pd.to_datetime(
    df['order_time'],
    format='%H:%M:%S'
)

df['day_hour'] = df['order_time'].dt.hour

# Count unique orders for each hour
orders_by_hour = df.groupby('day_hour')['order_id'].nunique()

# Plot bar chart
ax = orders_by_hour.plot(
    kind='bar',
    figsize=(10, 5),
    color='pink',
    edgecolor='black'
)

plt.title('Total Orders by Hour of the Day')
plt.xlabel('Hour of the Day (24-hour format)')
plt.ylabel('Total Number of Orders')
plt.xticks(rotation=0)

for i, value in enumerate(orders_by_hour):
    plt.text(i, value + 5, str(value),
             ha='center', va='bottom',
             fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python

# Convert order_date into datetime format
df['order_date'] = pd.to_datetime(
    df['order_date'],
    dayfirst=True
)

# Extract month name
df['month'] = df['order_date'].dt.month_name()

# Arrange months in calendar order
monthly_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

df['month'] = pd.Categorical(
    df['month'],
    categories=monthly_order,
    ordered=True
)

# Count unique orders for each month
orders_by_month = df.groupby(
    'month', observed=False
)['order_id'].nunique()

# Create bar chart
plt.figure(figsize=(10, 5))
plt.fill_between(orders_by_month.index, orders_by_month.values, color='lightblue', alpha=0.5)
plt.plot(orders_by_month.index, orders_by_month.values, color='blue', linewidth=2,marker='o')



# Add chart title and labels
plt.title('Total Orders by Month')
plt.xlabel('Month')
plt.ylabel('Total Number of Orders')
plt.xticks(rotation=45)

# Display values on top of each bar
for i, value in enumerate(orders_by_month):
    plt.text(
        i, value + 20,str(value),
        ha='center',
        va='bottom',
        fontsize=9,
        fontweight='bold'
    )

plt.tight_layout()
plt.show()
```

```python

category_sales = df.groupby('pizza_category')['total_price'].sum()

category_pct = category_sales / category_sales.sum() * 100

plt.figure(figsize=(7,7))
colors = plt.get_cmap('tab20').colors

plt.pie(category_pct, labels=category_pct.index,
        autopct='%1.1f%%', startangle=90,
        colors=colors,
        wedgeprops={'edgecolor':'black', 'width':0.5})

plt.title("Percentage of Sales by Pizza Category")
plt.show()
```

```python
sales_pivot = df.pivot_table(
    index='pizza_category',
    columns='pizza_size',
    values='total_price',
    aggfunc='sum',
    fill_value=0
)

seles_pct = sales_pivot/sales_pivot.sum().sum() * 100

plt.figure(figsize=(10, 6))
sns.heatmap(seles_pct, annot=True, fmt=".1f", cmap="YlGnBu", linewidths=0.5)
plt.title("Percentage of Sales by Pizza Category and Size")
plt.ylabel("Pizza Category")
plt.xlabel("Pizza Size")
plt.show()
```

```python

pizza_by_cartegory = df.groupby('pizza_category')['quantity'].sum()

colors = list(plt.get_cmap('tab20').colors)

colors = colors[:len(pizza_by_cartegory)]

ax = pizza_by_cartegory.plot(
    kind='bar',
    figsize=(8, 5),
    color=colors,
    edgecolor='black'
)

plt.title('Total Pizzas Sold by Category')
plt.xlabel('Pizza Category')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45)

for i, val in enumerate(pizza_by_cartegory):
    plt.text(i, val + 5, str(val),
             ha='center',
             va='bottom',
             fontsize=9,
             fontweight='bold')

plt.tight_layout()
plt.show()
```

```python

pizzas_by_name = df.groupby('pizza_name')['quantity'].sum()


top_5_pizzas = pizzas_by_name.sort_values(ascending=False).head(5)

ax = top_5_pizzas.plot(
    kind='bar',
    figsize=(10, 5),
    color=colors[:5],
    edgecolor='black'
)

plt.title('Top 5 Best Selling Pizzas')
plt.xlabel('Pizza Name')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45)

for i, val in enumerate(top_5_pizzas):
    plt.text(i, val + 2, str(val), ha='center', va='bottom', fontsize=9, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python

# Group pizzas by name and calculate total quantity
pizzas_by_name = df.groupby('pizza_name')['quantity'].sum()

# Get Top 5 best selling pizzas
top_5_pizzas = pizzas_by_name.sort_values(ascending=False).head(5)

# Display Top 5 pizzas
print("Top 5 Best Selling Pizzas:")
print(top_5_pizzas)

# Bar Chart
colors = list(plt.get_cmap('tab10').colors)

ax = top_5_pizzas.plot(
    kind='bar',
    figsize=(10, 5),
    color=colors[:5],
    edgecolor='black'
)

plt.title('Top 5 Best Selling Pizzas')
plt.xlabel('Pizza Name')
plt.ylabel('Total Quantity Sold')
plt.xticks(rotation=45, ha='right')

# Add quantity labels
for i, val in enumerate(top_5_pizzas):
    plt.text(i, val + 2, str(val),
             ha='center',
             va='bottom',
             fontsize=9,
             fontweight='bold')

plt.tight_layout()
plt.show()
```

```python

# Group pizzas by name and calculate total revenue
pizzas_by_name = df.groupby('pizza_name')['total_price'].sum()

# Get Top 5 pizzas by total sales
top_5_pizzas = pizzas_by_name.sort_values(ascending=False).head(5)

# Display Top 5 pizzas
print("Top 5 Best Selling Pizzas by Total Revenue:")
print(top_5_pizzas)

# Bar Chart
colors = list(plt.get_cmap('tab10').colors)

ax = top_5_pizzas.plot(
    kind='bar',
    figsize=(10, 5),
    color=colors[:5],
    edgecolor='black'
)

plt.title('Top 5 Pizzas by Total Revenue')
plt.xlabel('Pizza Name')
plt.ylabel('Total Revenue')
plt.xticks(rotation=45, ha='right')

# Add revenue labels
for i, val in enumerate(top_5_pizzas):
    plt.text(i, val + 2, str(round(val, 2)),
             ha='center',
             va='bottom',
             fontsize=9,
             fontweight='bold')

plt.tight_layout()
plt.show()
```
